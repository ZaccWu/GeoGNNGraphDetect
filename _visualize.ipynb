```python
# 生成x值
beta = np.linspace(1, 20, 400)  # 从-10到10生成400个点

def plot_curve(axes, beta, recall_list, rate, title):
    y1 = beta**recall_list[0] - 10*rate
    y2 = beta**recall_list[1] - 10*rate
    y3 = beta**recall_list[2] - 10*rate
    y4 = beta**recall_list[3] - 10*rate
    axes.plot(beta, y1, label='GEOM', linewidth=2.5, color='orangered')  
    axes.plot(beta, y2, label='StableGNN', linewidth=2.5, color='deeppink')  
    axes.plot(beta, y3, label='MoEGAD', linewidth=2.5, color='pink')  
    axes.plot(beta, y4, label='Random', linewidth=2.5, color='darkred') 
    axes.tick_params(labelsize=15) 
    axes.set_title(title, fontsize=18)  # 设置标题
    axes.grid(True)  # 显示网格线


fig, axes = plt.subplots(2, 6, figsize=(24, 8))

plot_curve(axes[0][0], beta, [0.0507, 0.0404, 0.0362, 0.0095], rate=0.01, title='Soc1 Recall@1%')
plot_curve(axes[0][1], beta, [0.0481, 0.0366, 0.0355, 0.0080], rate=0.01, title='Soc2 Recall@1%')
plot_curve(axes[0][2], beta, [0.0389, 0.0326, 0.0205, 0.0093], rate=0.01, title='Busi Recall@1%')
plot_curve(axes[0][3], beta, [0.1092, 0.0979, 0.0749, 0.0205], rate=0.02, title='Soc1 Recall@2%')
plot_curve(axes[0][4], beta, [0.0916, 0.0843, 0.0715, 0.0164], rate=0.02, title='Soc2 Recall@2%')
plot_curve(axes[0][5], beta, [0.0726, 0.0721, 0.0394, 0.0211], rate=0.02, title='Busi Recall@2%')
plot_curve(axes[1][0], beta, [0.2154, 0.1854, 0.1420, 0.0521], rate=0.05, title='Soc1 Recall@5%')
plot_curve(axes[1][1], beta, [0.1942, 0.1753, 0.1340, 0.0473], rate=0.05, title='Soc2 Recall@5%')
plot_curve(axes[1][2], beta, [0.1425, 0.1257, 0.0928, 0.0509], rate=0.05, title='Busi Recall@5%')
plot_curve(axes[1][3], beta, [0.3481, 0.3144, 0.2595, 0.0977], rate=0.1, title='Soc1 Recall@10%')
plot_curve(axes[1][4], beta, [0.3317, 0.3094, 0.2714, 0.0987], rate=0.1, title='Soc2 Recall@10%')
plot_curve(axes[1][5], beta, [0.2686, 0.2422, 0.1944, 0.0950], rate=0.1, title='Busi Recall@10%')
plt.legend(fontsize=16)

plt.tight_layout()   # 自动调整间距，防止重叠
plt.savefig('econ_sim.jpg', dpi=500)
plt.show()
```
